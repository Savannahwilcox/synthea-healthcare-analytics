```python
import pandas as pd
import numpy as np

# Load the drug pricing data
df = pd.read_csv('../data/raw/DSD_PTD_RY26_P04_V10_DY24_BGM.csv')

print(f"Dataset shape: {df.shape}")
print(f"\nColumn names:")
print(df.columns.tolist())
print(f"\nFirst few rows:")
print(df.head())
print(f"\nData types:")
print(df.dtypes)
```

```python
# Find drugs with biggest price increases
print("TOP 20 DRUGS WITH LARGEST PRICE INCREASES (2020-2024):")

# Sort by CAGR (compound annual growth rate) descending
top_increases = df.nlargest(20, 'CAGR_Avg_Spnd_Per_Dsg_Unt_20_24')[
    ['Brnd_Name', 'Gnrc_Name', 'Avg_Spnd_Per_Dsg_Unt_Wghtd_2020', 
     'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024', 'CAGR_Avg_Spnd_Per_Dsg_Unt_20_24']
]

# Rename for readability
top_increases.columns = ['Brand Name', 'Generic Name', 'Price 2020', 'Price 2024', 'Annual Growth Rate']
print(top_increases)

print("AVERAGE PRICE TREND ACROSS ALL DRUGS:")
print(f"2020: ${df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2020'].mean():.2f}")
print(f"2021: ${df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2021'].mean():.2f}")
print(f"2022: ${df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2022'].mean():.2f}")
print(f"2023: ${df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2023'].mean():.2f}")
print(f"2024: ${df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2024'].mean():.2f}")
```

```python
# Find drugs with biggest 2023-2024 increases
print("\nTOP 15 DRUGS WITH BIGGEST PRICE JUMPS (2023→2024):")

price_jump = df.nlargest(15, 'Chg_Avg_Spnd_Per_Dsg_Unt_23_24')[
    ['Brnd_Name', 'Gnrc_Name', 'Avg_Spnd_Per_Dsg_Unt_Wghtd_2023', 
     'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024', 'Chg_Avg_Spnd_Per_Dsg_Unt_23_24']
]
price_jump.columns = ['Brand', 'Generic', 'Price 2023', 'Price 2024', 'Change 2023-24']
print(price_jump)

# Count: how many drugs had price increases vs decreases?
print("\nPRICE DIRECTION SUMMARY:")
increases = (df['Chg_Avg_Spnd_Per_Dsg_Unt_23_24'] > 0).sum()
decreases = (df['Chg_Avg_Spnd_Per_Dsg_Unt_23_24'] < 0).sum()
print(f"Drugs with price INCREASES (2023-24): {increases}")
print(f"Drugs with price DECREASES (2023-24): {decreases}")
print(f"Percentage that increased: {(increases/(increases+decreases)*100):.1f}%")
```

```python
# What drugs cost the MOST in 2024?
print("\nTOP 20 MOST EXPENSIVE DRUGS (2024):")

expensive = df.nlargest(20, 'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024')[
    ['Brnd_Name', 'Gnrc_Name', 'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024', 
     'Tot_Clms_2024', 'Tot_Benes_2024']
]
expensive.columns = ['Brand', 'Generic', 'Price per Dose', 'Claims', 'Beneficiaries']
print(expensive)

print("\nTOTAL SPENDING BY TOP 20 DRUGS (2024):")

# Calculate total spending for each drug
df['Total_Spending_2024'] = df['Tot_Spndng_2024']

top_spending = df.nlargest(20, 'Tot_Spndng_2024')[
    ['Brnd_Name', 'Gnrc_Name', 'Tot_Spndng_2024', 'Tot_Benes_2024', 
     'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024']
]
top_spending.columns = ['Brand', 'Generic', 'Total Spending 2024', 'Beneficiaries', 'Price per Dose']
print(top_spending)
```

```python
# How many beneficiaries per drug?
print("\nDRUGS WITH MOST BENEFICIARIES (2024):")

volume = df.nlargest(20, 'Tot_Benes_2024')[
    ['Brnd_Name', 'Gnrc_Name', 'Tot_Benes_2024', 'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024', 
     'Tot_Spndng_2024']
]
volume.columns = ['Brand', 'Generic', 'Beneficiaries', 'Price per Dose', 'Total Spending']
volume['Total Spending'] = volume['Total Spending'].apply(lambda x: f"${x/1e9:.2f}B")
print(volume)

# What percentage of total spending goes to top drugs?
print("\nSPENDING CONCENTRATION:")
total_spending_all = df['Tot_Spndng_2024'].sum()
top_10_spending = df.nlargest(10, 'Tot_Spndng_2024')['Tot_Spndng_2024'].sum()
top_50_spending = df.nlargest(50, 'Tot_Spndng_2024')['Tot_Spndng_2024'].sum()

print(f"Total Medicare Part D Spending 2024: ${total_spending_all/1e9:.2f}B")
print(f"Top 10 drugs account for: ${top_10_spending/1e9:.2f}B ({top_10_spending/total_spending_all*100:.1f}%)")
print(f"Top 50 drugs account for: ${top_50_spending/1e9:.2f}B ({top_50_spending/total_spending_all*100:.1f}%)")
print(f"Remaining {len(df) - 50} drugs account for: ${(total_spending_all - top_50_spending)/1e9:.2f}B ({(total_spending_all - top_50_spending)/total_spending_all*100:.1f}%)")
```

```python
# Drugs by therapeutic category: which categories saw biggest price increases?
print("\nPRICE CHANGES BY DRUG TYPE:")

# Group by generic name prefix to identify drug types
df['Drug_Type'] = df['Gnrc_Name'].str.split(' ').str[0]  # First word = drug type

price_trends = df.groupby('Drug_Type').agg({
    'CAGR_Avg_Spnd_Per_Dsg_Unt_20_24': 'mean',
    'Chg_Avg_Spnd_Per_Dsg_Unt_23_24': 'mean',
    'Brnd_Name': 'count'
}).round(4)

price_trends.columns = ['Avg Annual Growth Rate', 'Change 2023-24', 'Number of Drugs']
price_trends = price_trends.sort_values('Avg Annual Growth Rate', ascending=False)
print(price_trends.head(20))
```

```python
# Manufacturer analysis 
print("\nTOP 20 MANUFACTURERS BY TOTAL SPENDING (2024):")

# Filter out "Overall" to get individual manufacturers only
df_mftr = df[df['Mftr_Name'] != 'Overall'].copy()

# Group by manufacturer and sum their spending
mftr_spending = df_mftr.groupby('Mftr_Name').agg({
    'Tot_Spndng_2024': 'sum',
    'Brnd_Name': 'count',
    'Tot_Benes_2024': 'sum'
}).round(2)

mftr_spending.columns = ['Total Spending 2024', 'Number of Drugs', 'Total Beneficiaries']
mftr_spending = mftr_spending.sort_values('Total Spending 2024', ascending=False)

# Format spending column
mftr_spending['Total Spending 2024'] = mftr_spending['Total Spending 2024'].apply(lambda x: f"${x/1e9:.2f}B")

print(mftr_spending.head(20))

print("\nMANUFACTURER MARKET CONCENTRATION:")

# Calculate concentration
total_spending = df_mftr['Tot_Spndng_2024'].sum()
top_5 = df_mftr.groupby('Mftr_Name')['Tot_Spndng_2024'].sum().nlargest(5).sum()
top_10 = df_mftr.groupby('Mftr_Name')['Tot_Spndng_2024'].sum().nlargest(10).sum()

print(f"Total manufacturers: {df_mftr['Mftr_Name'].nunique()}")
print(f"Top 5 manufacturers control: ${top_5/1e9:.2f}B ({top_5/total_spending*100:.1f}% of market)")
print(f"Top 10 manufacturers control: ${top_10/1e9:.2f}B ({top_10/total_spending*100:.1f}% of market)")
```

```python
print("ANALYSIS 2: PRICE vs BENEFICIARIES RELATIONSHIP")
print("=" * 80)

# Group drugs into price tiers
df['Price_Tier'] = pd.cut(df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2024'], 
                           bins=[0, 1, 10, 100, 1000, 1000000],
                           labels=['Under $1', '$1-10', '$10-100', '$100-1K', 'Over $1K'])

price_tier_analysis = df.groupby('Price_Tier', observed=True).agg({
    'Brnd_Name': 'count',
    'Tot_Benes_2024': 'mean',
    'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024': 'mean',
    'Tot_Spndng_2024': 'sum'
}).round(2)

price_tier_analysis.columns = ['Number of Drugs', 'Avg Beneficiaries per Drug', 'Avg Price per Dose', 'Total Spending']
price_tier_analysis['Total Spending'] = price_tier_analysis['Total Spending'].apply(lambda x: f"${x/1e9:.2f}B")

print(price_tier_analysis)

print("\nINSIGHT: Do expensive drugs have fewer patients?")
```

```python
import pandas as pd
import os

# Reload the data
df = pd.read_csv('../data/raw/DSD_PTD_RY26_P04_V10_DY24_BGM.csv')

# Recreate the Price_Tier column
df['Price_Tier'] = pd.cut(df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2024'], 
                           bins=[0, 1, 10, 100, 1000, 1000000],
                           labels=['Under $1', '$1-10', '$10-100', '$100-1K', 'Over $1K'])

# Recreate the Is_Generic column
df['Is_Generic'] = df['Brnd_Name'].str.lower() == df['Gnrc_Name'].str.lower()
df['Is_Generic'] = df['Is_Generic'] | df['Brnd_Name'].str.contains(r'\*|Generic', case=False, na=False)

print("Data reloaded")

# Create outputs folder
os.makedirs('../data/processed', exist_ok=True)

# 1. Price trend
price_trend = pd.DataFrame({
    'Year': [2020, 2021, 2022, 2023, 2024],
    'Avg_Price_per_Dose': [
        df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2020'].mean(),
        df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2021'].mean(),
        df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2022'].mean(),
        df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2023'].mean(),
        df['Avg_Spnd_Per_Dsg_Unt_Wghtd_2024'].mean()
    ]
})
price_trend.to_csv('../data/processed/price_trend.csv', index=False)
print("Price trend saved")

# 2. Top 20 drugs
top_drugs = df[df['Mftr_Name']=='Overall'].nlargest(20, 'Tot_Spndng_2024')[
    ['Brnd_Name', 'Gnrc_Name', 'Tot_Spndng_2024', 'Tot_Benes_2024', 
     'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024', 'CAGR_Avg_Spnd_Per_Dsg_Unt_20_24']
]
top_drugs.columns = ['Brand', 'Generic', 'Total_Spending_2024', 'Beneficiaries', 'Price_per_Dose', 'CAGR']
top_drugs.to_csv('../data/processed/top_20_drugs.csv', index=False)
print("Top 20 drugs saved")

# 3. Manufacturers
df_mftr = df[df['Mftr_Name'] != 'Overall'].copy()
mftr_data = df_mftr.groupby('Mftr_Name').agg({
    'Tot_Spndng_2024': 'sum',
    'Brnd_Name': 'count',
    'Tot_Benes_2024': 'sum'
}).reset_index()
mftr_data.columns = ['Manufacturer', 'Spending_2024', 'Number_of_Drugs', 'Total_Beneficiaries']
mftr_data = mftr_data.nlargest(20, 'Spending_2024')
mftr_data.to_csv('../data/processed/manufacturers.csv', index=False)
print("Manufacturers saved")

# 4. Price tiers
price_tiers = df.groupby('Price_Tier', observed=True).agg({
    'Brnd_Name': 'count',
    'Tot_Benes_2024': 'mean',
    'Tot_Spndng_2024': 'sum'
}).reset_index()
price_tiers.columns = ['Price_Tier', 'Number_of_Drugs', 'Avg_Beneficiaries', 'Total_Spending']
price_tiers.to_csv('../data/processed/price_tiers.csv', index=False)
print("Price tiers saved")

# 5. Generic vs brand
generic_brand = df.groupby('Is_Generic').agg({
    'Brnd_Name': 'count',
    'Avg_Spnd_Per_Dsg_Unt_Wghtd_2024': 'mean',
    'Tot_Benes_2024': 'mean',
    'Tot_Spndng_2024': 'sum'
}).reset_index()
generic_brand['Drug_Type'] = generic_brand['Is_Generic'].apply(lambda x: 'Generic' if x else 'Brand')
generic_brand.columns = ['Is_Generic', 'Number_of_Drugs', 'Avg_Price', 'Avg_Beneficiaries', 'Total_Spending', 'Drug_Type']
generic_brand = generic_brand[['Drug_Type', 'Number_of_Drugs', 'Avg_Price', 'Avg_Beneficiaries', 'Total_Spending']]
generic_brand.to_csv('../data/processed/generic_vs_brand.csv', index=False)
print("Generic vs brand saved")

print("All datasets ready for Tableau")
```
